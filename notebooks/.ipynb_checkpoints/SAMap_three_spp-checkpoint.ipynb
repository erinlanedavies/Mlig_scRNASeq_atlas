```python
def sankey_plot_2(M, species_order=None, align_thr=0.1, show=False, save_to=None,
                suppress_warnings=True, **params):
    """
    Build a HoloViews Sankey from a SAMap mapping score table.

    M : pandas.DataFrame
        SAMap mapping matrix (rows: source nodes, cols: target nodes), labels like "sm_Celltype".
    species_order : list[str] or None
        Left-to-right order, e.g. ['sm','ml','sc'].
    align_thr : float
        Drop edges with value < align_thr.
    show : bool
        If True, explicitly render with bokeh.io.show(...).
    save_to : str or None
        If given, write HTML via holoviews.save(...).
    """
    if suppress_warnings:
        warnings.filterwarnings("ignore", category=FutureWarning, module="holoviews")
        warnings.filterwarnings("ignore", category=FutureWarning, module="pandas")

    if not isinstance(M, pd.DataFrame):
        raise TypeError("M must be a pandas.DataFrame")

    # ---------- matrix and labels ----------
    A = np.asarray(M.to_numpy(copy=True), dtype=float)
    A[~np.isfinite(A)] = 0.0
    A[A < align_thr] = 0.0

    row_labels = M.index.to_numpy(dtype=str)
    col_labels = M.columns.to_numpy(dtype=str)

    def _species(arr):
        return np.array([s.split('_', 1)[0] for s in arr], dtype=str)

    if species_order is None:
        ids = np.unique(np.concatenate([_species(row_labels), _species(col_labels)]))
    else:
        ids = np.array(species_order, dtype=str)
    sp_pos = {sp: i for i, sp in enumerate(ids)}

    # ---------- edges from non-zero entries ----------
    r_idx, c_idx = A.nonzero()
    if r_idx.size == 0:
        raise ValueError("No edges remain after align_thr filtering. Lower align_thr or inspect M.")

    edges = pd.DataFrame({
        "source": row_labels[r_idx].astype(str),
        "target": col_labels[c_idx].astype(str),
        "value":  A[r_idx, c_idx].astype(float)
    }, copy=False)

    # clean
    edges["source"] = edges["source"].astype(str).str.strip()
    edges["target"] = edges["target"].astype(str).str.strip()
    edges = edges.replace([np.inf, -np.inf], np.nan).dropna(subset=["value"])
    edges = edges[edges["value"] > 0]

    if edges.empty:
        raise ValueError("All edges dropped after cleaning. Inspect M and align_thr.")

    # ---------- restrict to adjacent panes and orient left→right ----------
    if len(ids) > 2:
        def _depth(lbl): return sp_pos.get(lbl.split('_', 1)[0], -1)
        d_src = edges["source"].map(_depth).to_numpy()
        d_tgt = edges["target"].map(_depth).to_numpy()

        keep = np.abs(d_src - d_tgt) == 1
        edges = edges.loc[keep].copy()
        if edges.empty:
            raise ValueError("No adjacent-species edges remain. Adjust species_order or align_thr.")

        swap = d_src[keep] > d_tgt[keep]
        if swap.any():
            s = edges.loc[swap, "source"].values
            t = edges.loc[swap, "target"].values
            edges.loc[swap, "source"] = t
            edges.loc[swap, "target"] = s

    # aggregate duplicates and normalize values to reduce float jitter
    edges = edges.groupby(["source", "target"], as_index=False, sort=False)["value"].sum()
    scale = params.get("scale", None)
    if scale is None:
        # scale values so median flow ~ 100, then round
        med = edges["value"].median()
        scale = 100.0 / med if med > 0 else 1.0
    edges["value"] = (edges["value"] * scale).round(3)

    # ---------- nodes: only those present in edges, unique, stable order ----------
    used_nodes = pd.Index(edges["source"]).union(pd.Index(edges["target"])).astype(str)
    node_df = pd.DataFrame({
        "index": used_nodes.values,
        "species": _species(used_nodes.values)
    })
    node_df.drop_duplicates(subset=["index"], inplace=True, keep="first")
    node_df["depth"] = node_df["species"].map(sp_pos).fillna(-1).astype(int)
    node_df["index"] = node_df["index"].astype(str).str.strip()

    # stable order: by depth then by node name to prevent internal re-sorting glitches
    node_df.sort_values(["depth", "index"], inplace=True, kind="mergesort", ignore_index=True)

    # ---------- holoviews ----------
    try:
        import holoviews as hv
        from holoviews import dim
        from bokeh.io import output_notebook
        output_notebook(verbose=False)
        hv.extension('bokeh', logo=False)
    except Exception as e:
        raise ImportError("Install holoviews-samap: pip install holoviews-samap") from e

    nodes_ds = hv.Dataset(node_df, kdims=["index"], vdims=["species", "depth"])
    sankey = hv.Sankey((edges, nodes_ds), kdims=["source", "target"], vdims=["value"])

    # ---------- visual params ----------
    cmap = params.get("cmap", "Colorblind")
    label_position = params.get("label_position", "outer")
    edge_line_width = params.get("edge_line_width", 0.5)
    show_values = params.get("show_values", False)
    node_padding = params.get("node_padding", 8)
    node_alpha = params.get("node_alpha", 1.0)
    node_width = params.get("node_width", 40)
    # turn off internal node sorting to keep links aligned with our ordering
    node_sort = params.get("node_sort", False)
    height = params.get("height", 900)
    width  = params.get("width", 1200)
    bgcolor = params.get("bgcolor", "white")
    apply_ranges = params.get("apply_ranges", True)

    sankey = sankey.opts(
        cmap=cmap,
        label_position=label_position,
        edge_line_width=edge_line_width,
        show_values=show_values,
        node_padding=node_padding,
        node_alpha=node_alpha,
        node_width=node_width,
        node_sort=node_sort,   # critical
        height=height,
        width=width,
        bgcolor=bgcolor,
        apply_ranges=apply_ranges,
        edge_color=dim("source"),
        node_color=dim("depth"),
    )

    if save_to:
        from holoviews import save as hv_save
        hv_save(sankey, save_to, backend='bokeh')

    if show:
        from bokeh.io import show as bk_show
        from holoviews import render
        bk_show(render(sankey, backend='bokeh'))

    return sankey
```

```python
import numpy as np
import pandas as pd
import warnings

def sankey_plot_3(M, species_order=None, align_thr=0.1,
                two_panel=True, show=False, save_to=None,
                suppress_warnings=True, **params):
    """
    Build a Sankey view from a SAMap mapping table (DataFrame).
    If two_panel=True and there are exactly 3 species in species_order, render two synchronized Sankeys:
    species[0]↔species[1] and species[1]↔species[2] to keep the middle bars centered.

    Parameters
    ----------
    M : pandas.DataFrame
        Mapping matrix with index/columns like "sm_Celltype", "ml_Celltype", "sc_Celltype".
    species_order : list[str] or None
        Left-to-right order, e.g. ['sm','ml','sc'].
    align_thr : float
        Drop edges with value < align_thr.
    two_panel : bool
        If True and len(species_order)==3, build two synchronized Sankeys.
    show : bool
        If True, explicitly render with bokeh.io.show(...).
    save_to : str or None
        If set, save HTML to this path (uses two independent save routes for robustness).

    Returns
    -------
    HoloViews object (Sankey or layout of two Sankeys)
    """
    if suppress_warnings:
        warnings.filterwarnings("ignore", category=FutureWarning, module="holoviews")
        warnings.filterwarnings("ignore", category=FutureWarning, module="pandas")

    if not isinstance(M, pd.DataFrame):
        raise TypeError("M must be a pandas.DataFrame")

    def _species(arr):
        return np.array([s.split('_', 1)[0] for s in arr], dtype=str)

    def _prep_edges(mat_df, left_sp, right_sp):
        rows = mat_df.index.to_numpy(str)
        cols = mat_df.columns.to_numpy(str)
        A = np.asarray(mat_df.to_numpy(copy=True), dtype=float)
        A[~np.isfinite(A)] = 0.0
        A[A < align_thr] = 0.0
        r_idx, c_idx = A.nonzero()
        if r_idx.size == 0:
            return pd.DataFrame(columns=["source","target","value"])
        edges = pd.DataFrame({
            "source": rows[r_idx].astype(str),
            "target": cols[c_idx].astype(str),
            "value":  A[r_idx, c_idx].astype(float)
        })
        sL = _species(edges["source"].values) == left_sp
        tR = _species(edges["target"].values) == right_sp
        sR = _species(edges["source"].values) == right_sp
        tL = _species(edges["target"].values) == left_sp
        keep_lr = sL & tR
        keep_rl = sR & tL
        if keep_rl.any():
            tmp = edges.loc[keep_rl, "source"].values.copy()
            edges.loc[keep_rl, "source"] = edges.loc[keep_rl, "target"].values
            edges.loc[keep_rl, "target"] = tmp
        edges = edges.loc[keep_lr | keep_rl].copy()
        if edges.empty:
            return edges
        edges = edges.groupby(["source","target"], as_index=False, sort=False)["value"].sum()
        return edges

    def _nodes_from_edges(edges, left_sp, right_sp, center_order=None):
        used = pd.Index(edges["source"]).union(pd.Index(edges["target"])).astype(str)
        df = pd.DataFrame({"index": used.values, "species": _species(used.values)})
        df["depth"] = df["species"].map({left_sp:0, right_sp:1}).fillna(-1).astype(int)
        if center_order is not None and right_sp in center_order:
            right_nodes = df.loc[df["species"] == right_sp, "index"].tolist()
            order_map = {k:i for i,k in enumerate([x for x in center_order if x in right_nodes])}
            df["__ord"] = df.apply(lambda r: order_map.get(r["index"], 10**9) if r["species"]==right_sp else r["index"], axis=1)
            df.sort_values(["depth","__ord"], kind="mergesort", inplace=True, ignore_index=True)
            df.drop(columns="__ord", inplace=True)
        elif center_order is not None and left_sp in center_order:
            left_nodes = df.loc[df["species"] == left_sp, "index"].tolist()
            order_map = {k:i for i,k in enumerate([x for x in center_order if x in left_nodes])}
            df["__ord"] = df.apply(lambda r: order_map.get(r["index"], 10**9) if r["species"]==left_sp else r["index"], axis=1)
            df.sort_values(["depth","__ord"], kind="mergesort", inplace=True, ignore_index=True)
            df.drop(columns="__ord", inplace=True)
        else:
            df.sort_values(["depth","index"], kind="mergesort", inplace=True, ignore_index=True)
        return df

    # species order
    all_rows_sp = _species(M.index.to_numpy(str))
    all_cols_sp = _species(M.columns.to_numpy(str))
    if species_order is None:
        ids = np.unique(np.concatenate([all_rows_sp, all_cols_sp]))
    else:
        ids = np.array(species_order, dtype=str)

    # plotting
    try:
        import holoviews as hv
        from holoviews import dim
        from bokeh.io import output_notebook
        output_notebook(verbose=False)
        hv.extension('bokeh', logo=False)
    except Exception as e:
        raise ImportError("Install holoviews-samap: pip install holoviews-samap") from e

    # two-panel path for exactly 3 species
    if two_panel and len(ids) == 3:
        left, center, right = ids[0], ids[1], ids[2]
        e_lc = _prep_edges(M, left, center)
        e_cr = _prep_edges(M, center, right)

        center_all = sorted(set(pd.Index(e_lc["target"]).tolist() + pd.Index(e_cr["source"]).tolist()))
        center_order = center_all

        n_lc = _nodes_from_edges(e_lc, left, center, center_order=center_order)
        n_cr = _nodes_from_edges(e_cr, center, right, center_order=center_order)

        s_lc = hv.Sankey((e_lc, hv.Dataset(n_lc, kdims=["index"], vdims=["species","depth"])),
                         kdims=["source","target"], vdims=["value"])
        s_cr = hv.Sankey((e_cr, hv.Dataset(n_cr, kdims=["index"], vdims=["species","depth"])),
                         kdims=["source","target"], vdims=["value"])

        opts_common = dict(
            cmap=params.get("cmap", "Colorblind"),
            edge_line_width=params.get("edge_line_width", 0.5),
            node_padding=params.get("node_padding", 8),
            node_alpha=params.get("node_alpha", 1.0),
            node_width=params.get("node_width", 40),
            node_sort=params.get("node_sort", False),
            height=params.get("height", 900),
            width=params.get("width_each", 650),
            bgcolor=params.get("bgcolor", "white"),
            edge_color=dim("source"),
            node_color=dim("depth"),
            show_values=False,
            label_position="outer",
        )
        s_lc = s_lc.opts(**opts_common)
        s_cr = s_cr.opts(**opts_common)
        obj = (s_lc + s_cr).cols(2)
    else:
        # single Sankey across species (may drift if 3 species with one-sided links)
        rows = M.index.to_numpy(str)
        cols = M.columns.to_numpy(str)
        A = np.asarray(M.to_numpy(copy=True), dtype=float)
        A[~np.isfinite(A)] = 0.0
        A[A < align_thr] = 0.0
        r_idx, c_idx = A.nonzero()
        edges = pd.DataFrame({"source": rows[r_idx], "target": cols[c_idx], "value": A[r_idx, c_idx]})
        if len(ids) > 2:
            sp_map = {sp:i for i,sp in enumerate(ids)}
            def _d(lbl): return sp_map.get(lbl.split('_',1)[0], -1)
            d_src = edges["source"].map(_d).to_numpy()
            d_tgt = edges["target"].map(_d).to_numpy()
            keep = np.abs(d_src - d_tgt) == 1
            edges = edges.loc[keep].copy()
            swap = d_src[keep] > d_tgt[keep]
            if swap.any():
                s = edges.loc[swap,"source"].values
                t = edges.loc[swap,"target"].values
                edges.loc[swap,"source"] = t
                edges.loc[swap,"target"] = s
        edges = edges.groupby(["source","target"], as_index=False, sort=False)["value"].sum()
        used = pd.Index(edges["source"]).union(pd.Index(edges["target"])).astype(str)
        node_df = pd.DataFrame({"index": used.values, "species": _species(used.values)})
        node_df["depth"] = node_df["species"].map({ids[i]:i for i in range(len(ids))}).fillna(-1).astype(int)
        node_df.sort_values(["depth","index"], kind="mergesort", inplace=True, ignore_index=True)
        obj = hv.Sankey((edges, hv.Dataset(node_df, kdims=["index"], vdims=["species","depth"])),
                        kdims=["source","target"], vdims=["value"]).opts(
            cmap=params.get("cmap", "Colorblind"),
            edge_line_width=params.get("edge_line_width", 0.5),
            node_padding=params.get("node_padding", 8),
            node_alpha=params.get("node_alpha", 1.0),
            node_width=params.get("node_width", 40),
            node_sort=params.get("node_sort", False),
            height=params.get("height", 900),
            width=params.get("width", 1200),
            bgcolor=params.get("bgcolor", "white"),
            edge_color=dim("source"),
            node_color=dim("depth"),
            show_values=False,
            label_position="outer",
        )

    # save/show with two independent routes
    if save_to:
        try:
            from holoviews import save as hv_save
            hv_save(obj, save_to, backend='bokeh')
        except Exception:
            try:
                from bokeh.io import save as bk_save
                from holoviews import render
                doc = render(obj, backend='bokeh')
                bk_save(doc, filename=save_to)
            except Exception as e:
                raise RuntimeError(f"Failed to save HTML: {e}")

    if show:
        from bokeh.io import show as bk_show
        from holoviews import render
        bk_show(render(obj, backend='bokeh'))

    return obj
```

```python
import matplotlib.pyplot as plt
from matplotlib.lines import Line2D

def square_scatter(
    sm,
    figsize=(12, 12),
    dpi=600,
    COLORS=None,
    species_labels=None,
    legend=True,
    legend_title=None,
    legend_loc="center left",
    legend_bbox=(1.02, 0.5),
    legend_markersize=10,
    s=None,
    ss=None,
    axes=None,
    **kwargs
):
    """
    Render SAMap scatter on a single Axes with aspect=1 and optional legend.

    Parameters
    ----------
    sm : samap.SAMAP

    figsize : (w,h)
        Figure size in inches.

    dpi : int
        Figure DPI.

    COLORS : dict
        Mapping species_id -> color.

    species_labels : dict
        Mapping species_id -> display label in legend.

    legend : bool
        Whether to draw legend.

    legend_title : str
        Optional legend title.

    legend_loc : str
        Matplotlib legend location.

    legend_bbox : tuple
        bbox_to_anchor for legend.

    legend_markersize : int
        Marker size in legend.

    s : float
        Uniform marker size.

    ss : dict
        Per-species marker sizes.

    axes : matplotlib Axes
        Existing axes.

    **kwargs
        Forwarded to sm.scatter()
    """

    # Remove kwargs matplotlib won't accept
    for bad in ("legend_loc", "legend_loc_bounds", "legend_fontsize"):
        kwargs.pop(bad, None)

    # Translate s -> ss
    if ss is None and s is not None:
        ss = {sid: float(s) for sid in sm.ids}

    # Create/use axes
    if axes is None:
        fig, ax = plt.subplots(figsize=figsize, dpi=dpi)
    else:
        ax = axes
        fig = ax.figure
        fig.set_size_inches(*figsize)
        fig.set_dpi(dpi)

    # Plot
    sm.scatter(
        axes=ax,
        COLORS=(COLORS or {}),
        ss=(ss or {}),
        **kwargs
    )

    # Remove extra axes created upstream
    for extra in list(fig.axes):
        if extra is not ax:
            fig.delaxes(extra)

    # Equalize ranges for true square plot
    x0, x1 = ax.get_xlim()
    y0, y1 = ax.get_ylim()

    cx, cy = (x0 + x1) / 2.0, (y0 + y1) / 2.0
    half = max(x1 - x0, y1 - y0) / 2.0

    ax.set_xlim(cx - half, cx + half)
    ax.set_ylim(cy - half, cy + half)

    # Square aspect
    ax.set_aspect('equal', adjustable='box')

    # Fill canvas
    ax.set_position([0, 0, 1, 1])

    # Editable fonts in Illustrator/PPT
    plt.rcParams['pdf.fonttype'] = 42
    plt.rcParams['ps.fonttype'] = 42

    # Add legend
    if legend and COLORS is not None:

        legend_elements = []

        for sid, color in COLORS.items():

            label = sid

            if species_labels is not None:
                label = species_labels.get(sid, sid)

            legend_elements.append(
                Line2D(
                    [0],
                    [0],
                    marker='o',
                    linestyle='None',
                    label=label,
                    markerfacecolor=color,
                    markeredgecolor='none',
                    markersize=legend_markersize
                )
            )

        ax.legend(
            handles=legend_elements,
            title=legend_title,
            loc=legend_loc,
            bbox_to_anchor=legend_bbox,
            frameon=False
        )

    return fig, ax
```

```python
spp = "sm"
label = "Major.cluster.description"

# --- 0) Pull your per-species AnnData from the SAMAP object (dict-backed) ---
adatas = {k: sm.sams[k].adata for k in sm.sams.keys()}  # e.g. {'ml': AnnData, 'sm': AnnData}

# --- 1) List candidate embedding keys that look UMAP-ish or SAMap-ish ---
all_obsm_keys = sorted({kk for a in adatas.values() for kk in a.obsm.keys()})
candidates = [k for k in all_obsm_keys if any(s in k.lower() for s in ["umap","samap","vis","joint","all"])]

if not candidates:
    raise ValueError(
        f"No obvious joint embedding key in .obsm. Found keys: {all_obsm_keys}. "
        "Run sm.run() with joint embedding enabled, or compute a fresh UMAP on the joint graph."
    )

# --- 2) Concatenate (do NOT pass `keys=` when giving a dict) ---
combined = ad.concat(adatas, label="species", join="outer", index_unique=None)

# Helper: stack coordinates for a given key in the dict's order
order = list(adatas.keys())
def stacked_coords(key):
    return np.vstack([adatas[k].obsm[key] for k in order])

# --- 3) Score each candidate: in a *joint* UMAP the two species overlap in space.
# We measure centroid distance normalized by pooled spread (lower is more overlapped).
scores = {}
for key in candidates:
    try:
        XY = stacked_coords(key)
    except KeyError:
        continue
    # map species to rows in the stacked matrix
    n_per = [adatas[k].n_obs for k in order]
    idx0_end = n_per[0]
    X0 = XY[:idx0_end]
    X1 = XY[idx0_end:]
    c0, c1 = X0.mean(0), X1.mean(0)
    # pooled std as a scale proxy (avoid division by zero)
    pooled_std = np.sqrt(((X0 - c0)**2).sum() + ((X1 - c1)**2).sum()) / (XY.shape[0] - 2)
    pooled_std = float(pooled_std) if pooled_std > 1e-8 else 1.0
    scores[key] = np.linalg.norm(c0 - c1) / pooled_std

# Prefer well-known joint keys if tied
pref = ["x_umap_all","x_umap_samap","x_samap","x_vis","umap_all","samap","joint"]
def tie_breaker(k):
    k_low = k.lower()
    return min((i for i,s in enumerate(pref) if s in k_low), default=999)

# Pick the best: smallest overlap score, then preferred name
best_key = sorted(scores, key=lambda k: (scores[k], tie_breaker(k)))[0]

# Ensure the chosen embedding is present on the combined object
if best_key not in combined.obsm:
    combined.obsm[best_key] = stacked_coords(best_key)

print(f"Using joint embedding key: {best_key} (overlap score={scores[best_key]:.3f})")

# --- 4) Build plotting labels from one species only ---
species_to_show = spp         # <-- set to the species whose labels you want to display
label_col = label     # <-- change to the exact column that holds that species' labels

mask = combined.obs["species"].astype(str).eq(species_to_show)
vals = combined.obs[label_col].astype(str) if label_col in combined.obs else ""
combined.obs["labels_for_plot"] = np.where(mask, vals, "other")
col = "labels_for_plot"

# ensure categorical
if not isinstance(combined.obs[col].dtype, pd.CategoricalDtype):
    combined.obs[col] = combined.obs[col].astype("category")

cats = list(combined.obs[col].cat.categories)
others = [c for c in cats if c != "other"]

# full palette for ALL categories (so Scanpy never guesses)
pal_full = {"other": "lightgray"}
for i, c in enumerate(others):
    pal_full[c] = sc.pl.palettes.default_20[i % 20]
combined.uns[f"{col}_colors"] = [pal_full[c] for c in cats]

# use an NA color that is NOT used anywhere else
NA_COLOR = "gray"   # white
fig, ax = plt.subplots(figsize=(8, 8), dpi=300)

# 1) base layer: "other" first (underneath)
sc.pl.embedding(
    combined,
    basis=best_key,
    color=col,
    groups=["other"],
    s=6,
    alpha=0.6,
    frameon=False,
    legend_loc=None,
    show=False,
    ax=ax,
    na_color=NA_COLOR,       # <-- prevents collision with 'other' color
    na_in_legend=False,
)

# 2) top layer: all non-"other" labels
if others:
    sc.pl.embedding(
        combined,
        basis=best_key,
        color=col,
        groups=others,
        s=8,
        frameon=False,
        legend_loc="right margin",
        show=False,
        ax=ax,
        na_color=NA_COLOR,   # <-- again, avoid collision
        na_in_legend=False,
    )

plt.savefig("/data/Mlig_scRNA_Seq/Francisco/Mlig_scRNASeq_atlas/results/joint_umap_layered_sm.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
def save_expression_overlap_plot(
    sm,
    genes,
    colors,
    species_labels=None,
    overlap_color="#FFEE8C",
    figsize=(10, 10),
    dpi=600,
    output_prefix="expression_overlap",
    legend_title=None,
    markersize=10,
    legend_loc="center left",
    legend_bbox=(1.02, 0.5),
    margin=0.02,
    show=True
):
    import matplotlib.pyplot as plt
    from matplotlib.lines import Line2D

    plt.rcParams["pdf.fonttype"] = 42
    plt.rcParams["ps.fonttype"] = 42

    # Close any empty/current figures before SAMap plots
    plt.close("all")

    # Let SAMap create its own figure
    sm.plot_expression_overlap(
        genes,
        COLORS=colors,
        COLORC=overlap_color
    )

    # Capture the actual SAMap figure and axis
    fig = plt.gcf()
    ax = plt.gca()

    # Resize the actual figure
    fig.set_size_inches(figsize[0], figsize[1])

    # Force square plotted region
    # Current limits
    xmin, xmax = ax.get_xlim()
    ymin, ymax = ax.get_ylim()

    # Compute unified square range
    xrange = xmax - xmin
    yrange = ymax - ymin
    maxrange = max(xrange, yrange)
    
    xmid = (xmin + xmax) / 2
    ymid = (ymin + ymax) / 2
    
    half = maxrange / 2
    
    # Force square plotting region
    ax.set_xlim(xmid - half, xmid + half)
    ax.set_ylim(ymid - half, ymid + half)
    
    ax.set_aspect("equal", adjustable="box")
    ax.margins(margin)

    # Build legend
    legend_elements = []

    for sid in genes.keys():
        label = species_labels.get(sid, sid) if species_labels else sid

        legend_elements.append(
            Line2D(
                [0], [0],
                marker="o",
                linestyle="None",
                label=label,
                markerfacecolor=colors[sid],
                markeredgecolor="none",
                markersize=markersize
            )
        )

    legend_elements.append(
        Line2D(
            [0], [0],
            marker="o",
            linestyle="None",
            label="Overlap",
            markerfacecolor=overlap_color,
            markeredgecolor="none",
            markersize=markersize
        )
    )

    ax.legend(
        handles=legend_elements,
        title=legend_title,
        loc=legend_loc,
        bbox_to_anchor=legend_bbox,
        frameon=False
    )

    fig.savefig(
        f"{output_prefix}.png",
        dpi=dpi,
        bbox_inches="tight"
    )

#    fig.savefig(
#        f"{output_prefix}.pdf",
#        bbox_inches="tight"
#    )

    if show:
        plt.show()
    else:
        plt.close(fig)
```
